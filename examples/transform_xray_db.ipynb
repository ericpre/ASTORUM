<div>Transform original ESPM X-ray .json file into a format supported by BlockWiseSmoothNMF</div>

```python
from BlockWiseSmoothNMFModule.config import *
import json
```

```python
def transformXRayDatabase(original_json_file : str, transformed_json_file : str) -> None:
    with open(original_json_file, 'r') as f:
        xrays_data = json.load(f)

    with open(PERIODIC_TABLE_NUMBERS, 'r') as f:
        periodic_table_data = json.load(f)

    with open(SIEGBAHN_TO_IUPAC, 'r') as f:
        siegbahn_data = json.load(f)

    iupac_to_siegbahn = {}
    
    for siegbahn, iupac_list in siegbahn_data.items():
        if len(iupac_list) == 1:
            iupac_to_siegbahn[iupac_list[0]] = siegbahn

    atomic_number_to_symbol = {str(k): v['symbol'] for k, v in periodic_table_data['table'].items()}

    transformed_data = {'table': {}}
    for atomic_number, transitions in xrays_data['table'].items():
        symbol = atomic_number_to_symbol[atomic_number]
        transformed_data['table'][symbol] = {}
        for transition, data in transitions.items():
            siegbahn_transition = iupac_to_siegbahn.get(transition, transition)
            transformed_data['table'][symbol][siegbahn_transition] = data

    with open(transformed_json_file, 'w') as f:
        json.dump(transformed_data, f, indent = 4)
```
