```python
import json

# Load the JSON files
with open('/home/scozma/Documents/Python_scripts/Nickelate_superlattice/200kV_velox_new.json', 'r') as f:
    xrays_data = json.load(f)

with open('/home/scozma/Documents/Python_scripts/itSmoothNMF/C++/eigen/dev/periodic_table_number.json', 'r') as f:
    periodic_table_data = json.load(f)

with open('/home/scozma/Documents/Python_scripts/itSmoothNMF/C++/eigen/dev/siegbahn_to_iupac.json', 'r') as f:
    siegbahn_data = json.load(f)

# Create a reverse mapping from IUPAC to Siegbahn notation
iupac_to_siegbahn = {}
for siegbahn, iupac_list in siegbahn_data.items():
    if len(iupac_list) == 1:  # Only consider single mappings
        iupac_to_siegbahn[iupac_list[0]] = siegbahn

# Create a mapping from atomic number to symbol
atomic_number_to_symbol = {str(k): v['symbol'] for k, v in periodic_table_data['table'].items()}

# Transform the 200keV_xrays data
transformed_data = {'table': {}}
for atomic_number, transitions in xrays_data['table'].items():
    symbol = atomic_number_to_symbol[atomic_number]
    transformed_data['table'][symbol] = {}
    for transition, data in transitions.items():
        siegbahn_transition = iupac_to_siegbahn.get(transition, transition)  # Use original if no mapping found
        transformed_data['table'][symbol][siegbahn_transition] = data

# Save the transformed data back to a JSON file
with open('/home/scozma/Documents/Python_scripts/Nickelate_superlattice/200kV_velox_new_transformed.json', 'w') as f:
    json.dump(transformed_data, f, indent=4)
```
